## Part 1: Evolution of Kidney Systems and Ducts

### The Three Sequential Kidneys

Vertebrates develop THREE kidney systems during evolution and ontogeny:

**1. Pronephros** (First kidney)
- Most anterior position
- Functional only in primitive fish (hagfish)
- Transient in embryos of other vertebrates
- Creates the primary duct (archinephric duct)

**2. Mesonephros** (Middle kidney)
- Develops posterior to pronephros
- Uses same duct (now called mesonephric/Wolffian duct)
- Functional adult kidney in fish and amphibians
- Embryonic only in amniotes (reptiles, birds, mammals)

**3. Metanephros** (Final kidney)
- Most posterior position
- Creates NEW duct (ureter)
- Definitive kidney in all amniotes
- Allows separation of urinary and reproductive functions

### Key Innovation
The metanephros develops a **completely new duct** (ureter), which frees the mesonephric duct to be repurposed for **reproduction only**!

## Part 5: Types of Mammalian Uteri

### The Four Main Types

All mammalian uteri form from the fusion of TWO Müllerian ducts.  
The **degree of fusion** creates four distinct types:

**1. DUPLEX** - Completely separate (no fusion)
- Two complete uteri, two cervices
- Found in: Rodents, marsupials, elephants
- Litter size: 6-16 offspring

**2. BIPARTITE** - Partially fused body
- Large separate horns, small fused body, one cervix
- Found in: Carnivores (dogs, cats), pigs
- Litter size: 4-12 offspring

**3. BICORNUATE** - Small horns, large body
- Small horns, large dominant body, one cervix
- Found in: Ungulates (horses, cattle), whales
- Litter size: 1-3 offspring

**4. SIMPLEX** - Complete fusion (no horns)
- Single pear-shaped body, no horns, one cervix
- Found in: Higher primates (humans, apes), some bats
- Litter size: 1 offspring typical

## Part 7: Evolutionary Trends

```python
# Evolutionary trends in uterine fusion
def plot_evolutionary_trends():
    """
    Show correlation between fusion and reproductive traits
    """
    types = ['Duplex', 'Bipartite', 'Bicornuate', 'Simplex']
    fusion = [0, 25, 75, 100]
    litter_avg = [11, 8, 2, 1]  # Average litter sizes
    gestation_avg = [25, 85, 245, 210]  # Average gestation (days)
    
    fig = make_subplots(
        rows=1, cols=2,
        subplot_titles=('Fusion vs Litter Size', 'Fusion vs Gestation Length')
    )
    
    # Litter size
    fig.add_trace(go.Scatter(
        x=fusion,
        y=litter_avg,
        mode='lines+markers+text',
        text=types,
        textposition='top center',
        marker=dict(size=12, color='red'),
        line=dict(color='red', width=3),
        name='Litter Size'
    ), row=1, col=1)
    
    # Gestation length
    fig.add_trace(go.Scatter(
        x=fusion,
        y=gestation_avg,
        mode='lines+markers+text',
        text=types,
        textposition='top center',
        marker=dict(size=12, color='blue'),
        line=dict(color='blue', width=3),
        name='Gestation'
    ), row=1, col=2)
    
    fig.update_xaxes(title_text="Degree of Fusion (%)", row=1, col=1)
    fig.update_xaxes(title_text="Degree of Fusion (%)", row=1, col=2)
    fig.update_yaxes(title_text="Avg Litter Size", row=1, col=1)
    fig.update_yaxes(title_text="Gestation (days)", row=1, col=2)
    
    fig.update_layout(height=500, title_text='<b>Evolutionary Trends in Uterine Structure</b>',
                     showlegend=False)
    
    print("\nEVOLUTIONARY TRENDS")
    print("="*70)
    print("\nAS FUSION INCREASES (Duplex → Simplex):")
    print("  ↓ Litter size DECREASES (11 → 1 offspring)")
    print("  ↑ Gestation INCREASES (25 → 210+ days)")
    print("  ↑ Offspring size INCREASES")
    print("  ↑ Parental investment INCREASES")
    print("\nREPRODUCTIVE STRATEGY SHIFT:")
    print("  Duplex/Bipartite: r-selection (many small offspring, short life)")
    print("  Bicornuate/Simplex: K-selection (few large offspring, long life)")
    print("\nTRADE-OFF:")
    print("  Quantity vs Quality")
    print("  High reproductive rate vs High survival rate")
    print("="*70)
    
    fig.show()

display(HTML("<h3>📈 Evolutionary Correlations</h3>"))
plot_evolutionary_trends()
```

## Summary

### Part 1: Urinogenital Duct Evolution

**Three Sequential Kidneys:**
1. **Pronephros** - First kidney, creates archinephric duct
2. **Mesonephros** - Middle kidney, uses same duct (now Wolffian)
3. **Metanephros** - Final kidney, creates NEW duct (ureter)

**Key Innovation:**
- Metanephros creates separate ureter for urine
- Frees Wolffian/Müllerian ducts for reproduction only
- Allows independent evolution of urinary and reproductive systems

**Sex-Specific Differentiation:**

**Males:**
- Wolffian duct → Vas deferens (PERSISTS)
- Müllerian duct → Degenerates (vestigial)
- Function: Sperm transport

**Females:**
- Wolffian duct → Degenerates (vestigial)
- Müllerian duct → Oviduct, Uterus, Vagina (PERSISTS)
- Function: Egg transport, gestation

**Both:**
- Ureter → Kidney to bladder (PERSISTS)
- Function: Urine transport

### Part 2: Mammalian Uterine Types

**Four Types (by Müllerian duct fusion):**

| Type | Fusion | Horns | Body | Cervix | Litter | Examples |
|------|--------|-------|------|--------|--------|----------|
| **Duplex** | 0% | Two complete | None | Two | 6-16 | Rodents, marsupials |
| **Bipartite** | 25% | Two large | Small | One | 4-12 | Carnivores, pigs |
| **Bicornuate** | 75% | Two small | Large | One | 1-3 | Ungulates, whales |
| **Simplex** | 100% | None | Large | One | 1 | Primates, bats |

**Evolutionary Trends:**
- **More fusion** → Fewer offspring → Longer gestation → Larger offspring
- **Reproductive strategy shift:** r-selection (quantity) → K-selection (quality)
- **Trade-off:** Many small vs Few large offspring

### Key Principles

**Homology:**
- Same embryonic structures → Different adult forms
- Wolffian duct: Vas deferens (M) vs Vestigial (F)
- Müllerian duct: Vestigial (M) vs Uterus (F)

**Repurposing:**
- Kidney ducts repurposed for reproduction
- New kidney (metanephros) freed old ducts
- Allows specialization of both systems

**Developmental Basis:**
- Degree of fusion determines uterine type
- Human anomalies recapitulate other mammalian types
- Developmental errors reveal evolutionary history

### Clinical Significance

**Vestigial Structures:**
- Gartner's duct (F) - Wolffian remnant
- Prostatic utricle (M) - Müllerian remnant
- Can form cysts, cause pathology

**Human Uterine Anomalies:**
- 4-7% of women have developmental errors
- Septate most common (~35% of anomalies)
- Can affect fertility and pregnancy outcomes
- Recapitulate other mammalian uterine types

### The Big Picture

**Evolution repurposes existing structures:**
- Kidney ducts become reproductive ducts
- New kidney (metanephros) allows this specialization
- Sex-specific selection shapes male vs female systems

**Development recapitulates evolution:**
- Three kidneys develop sequentially in embryo
- Reflects evolutionary history
- Errors produce "throwback" structures

**Form follows function:**
- Uterine type matches reproductive strategy
- Many small offspring = Duplex
- Single large offspring = Simplex
- Structure optimized for life history

**Pattern recognition across vertebrates:**
- Same ducts, different uses
- Sequential kidney evolution
- Spectrum of uterine fusion
- Homology unites diversity
