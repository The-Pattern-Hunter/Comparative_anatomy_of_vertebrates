## Part 1: Alimentary Canal Overview

### Complete Digestive Tract (Mouth → Anus)

All vertebrates have a **complete digestive system** with:
- One-way flow
- Regional specialization
- Associated glands

### Seven Major Regions

**1. ORAL CAVITY** - Mechanical breakdown, initial digestion  
**2. PHARYNX** - Swallowing  
**3. ESOPHAGUS** - Transport via peristalsis  
**4. STOMACH** - Storage, acid digestion, protein breakdown  
**5. SMALL INTESTINE** - PRIMARY digestion and absorption (90% of nutrients!)  
**6. LARGE INTESTINE** - Water absorption, bacterial fermentation  
**7. CLOACA or ANUS** - Waste elimination

### Four Tissue Layers (Inside → Outside)

**1. MUCOSA** - Innermost, epithelium for absorption/secretion  
**2. SUBMUCOSA** - Blood vessels, nerves, connective tissue  
**3. MUSCULARIS EXTERNA** - Two muscle layers for peristalsis  
**4. SEROSA/ADVENTITIA** - Outermost protective layer

## Part 2: Associated Digestive Glands

```python
# Complete digestive glands database
digestive_glands = {
    'Salivary Glands': {
        'location': '3 pairs: Parotid, Submandibular, Sublingual',
        'secretion': 'Saliva (water, mucus, enzymes)',
        'enzymes': 'Salivary amylase (starch→maltose), Lingual lipase (fat)',
        'volume': '1-1.5 L/day',
        'pH': '6.5-7.5',
        'function': 'Lubrication, initial starch digestion, antimicrobial'
    },
    'Gastric Glands': {
        'location': 'Stomach wall (mucosa)',
        'secretion': 'Gastric juice (HCl, pepsinogen, mucus)',
        'enzymes': 'Pepsin (protein→peptides), Gastric lipase',
        'volume': '2-3 L/day',
        'pH': '1.5-3.5',
        'function': 'Protein digestion, acid environment, kill bacteria'
    },
    'Liver': {
        'location': 'Upper right abdomen (LARGEST gland)',
        'secretion': 'Bile (bile salts, bilirubin, cholesterol)',
        'enzymes': 'None (bile emulsifies, not enzymatic)',
        'volume': '0.5-1 L bile/day',
        'pH': '7.6-8.6 (alkaline)',
        'function': 'Fat emulsification, detoxification, metabolism'
    },
    'Gallbladder': {
        'location': 'Beneath liver',
        'secretion': 'Concentrated bile (from liver)',
        'enzymes': 'None (storage organ)',
        'volume': '30-50 mL capacity',
        'pH': '7.6-8.6',
        'function': 'Bile storage and concentration (up to 10x)'
    },
    'Pancreas': {
        'location': 'Behind stomach (exocrine 99% + endocrine 1%)',
        'secretion': 'Pancreatic juice (enzymes, bicarbonate)',
        'enzymes': 'Trypsin, Chymotrypsin, Amylase, Lipase, Nucleases',
        'volume': '1.5-2 L/day',
        'pH': '8.0-8.3 (very alkaline)',
        'function': 'Digestion of ALL nutrients, neutralizes stomach acid'
    },
    'Intestinal Glands': {
        'location': 'Small intestine (crypts of Lieberkühn)',
        'secretion': 'Intestinal juice (succus entericus)',
        'enzymes': 'Maltase, Lactase, Sucrase, Peptidases',
        'volume': '2-3 L/day',
        'pH': '7.5-8.0',
        'function': 'Final digestion at brush border'
    }
}

# Summary table
gland_data = []
for gland, data in digestive_glands.items():
    gland_data.append({
        'Gland': gland,
        'Volume': data['volume'],
        'pH': data['pH'],
        'Key_Function': data['function'][:40] + '...'
    })

df_glands = pd.DataFrame(gland_data)

print("\nDIGESTIVE GLANDS")
print("="*100)
print(df_glands.to_string(index=False))
print("\n" + "="*100)
print("TOTAL DAILY SECRETION: 8-10 LITERS!")
print("\nMAJOR SECRETIONS:")
print("  • Saliva: 1-1.5 L (amylase for starch)")
print("  • Gastric juice: 2-3 L (HCl pH 1.5-3.5, pepsin)")
print("  • Bile: 0.5-1 L (fat emulsification)")
print("  • Pancreatic juice: 1.5-2 L (enzymes for ALL nutrients, pH 8.0-8.3)")
print("  • Intestinal juice: 2-3 L (brush border enzymes)")
print("\nPANCREAS IS KEY: Neutralizes stomach acid + digests proteins, carbs, fats, nucleic acids!")
print("="*100)
```

## Part 5: Interactive Digestive System Explorer

```python
def explore_digestive(animal_type):
    """
    Interactive explorer for digestive systems
    """
    data = comparative_systems[animal_type]
    
    # Create display
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Diet & Teeth', 'Stomach', 'Intestines', 'Summary'),
        specs=[[{'type': 'table'}, {'type': 'table'}],
               [{'type': 'table'}, {'type': 'table'}]]
    )
    
    # Diet & Teeth
    fig.add_trace(go.Table(
        header=dict(values=['<b>Diet & Teeth</b>'], fill_color='lightblue'),
        cells=dict(values=[[f"Diet: {data['diet']}", f"Teeth: {data['teeth']}"]],
                  fill_color='white', align='left')
    ), row=1, col=1)
    
    # Stomach
    fig.add_trace(go.Table(
        header=dict(values=['<b>Stomach</b>'], fill_color='lightgreen'),
        cells=dict(values=[[data['stomach']]],
                  fill_color='white', align='left')
    ), row=1, col=2)
    
    # Intestines
    fig.add_trace(go.Table(
        header=dict(values=['<b>Intestines</b>'], fill_color='pink'),
        cells=dict(values=[[f"Length: {data['intestine_length']}",
                           f"Cecum: {data['cecum']}"]],
                  fill_color='white', align='left')
    ), row=2, col=1)
    
    # Summary
    fig.add_trace(go.Table(
        header=dict(values=['<b>Feature</b>', '<b>Value</b>'], fill_color='lightyellow'),
        cells=dict(values=[['Transit Time', 'Cellulose'],
                          [data['transit_time'], data['cellulose']]],
                  fill_color='white', align='left')
    ), row=2, col=2)
    
    fig.update_layout(height=500, title_text=f'<b>{animal_type}</b>')
    
    print("\n" + "="*90)
    print(f"{animal_type.upper()}")
    print("="*90)
    print(f"Diet: {data['diet']}")
    print(f"Teeth: {data['teeth']}")
    print(f"Stomach: {data['stomach']}")
    print(f"Intestine: {data['intestine_length']}")
    print(f"Transit: {data['transit_time']}")
    print(f"Cellulose: {data['cellulose']}")
    print("="*90)
    
    fig.show()

display(HTML("<h3>🍽️ Digestive System Explorer</h3>"))
interact(explore_digestive,
        animal_type=Dropdown(
            options=list(comparative_systems.keys()),
            value='Omnivore (Human/Pig)',
            description='Type:'
        ));
```

## Summary

### Alimentary Canal (7 Regions)

**1. Oral Cavity** - Chewing, saliva (amylase)  
**2. Pharynx** - Swallowing  
**3. Esophagus** - Transport (no digestion)  
**4. Stomach** - Acid (pH 1.5-3.5), pepsin  
**5. Small Intestine** - 90% absorption!  
**6. Large Intestine** - Water absorption  
**7. Anus/Cloaca** - Elimination

### Digestive Glands (6 Major)

**Total: 8-10 L/day secreted!**

1. **Salivary** (1-1.5 L) - Amylase  
2. **Gastric** (2-3 L) - HCl, pepsin  
3. **Liver** (0.5-1 L) - Bile (fat emulsification)  
4. **Gallbladder** - Stores bile (10x concentration)  
5. **Pancreas** (1.5-2 L) - ALL enzymes, pH 8.0-8.3  
6. **Intestinal** (2-3 L) - Brush border enzymes

### Stomach Types

**Simple** - 1 chamber (carnivores, omnivores)  
**Ruminant** - 4 chambers (Rumen, Reticulum, Omasum, Abomasum)  
**Bird** - 2 parts (Proventriculus + Gizzard)  
**Absent** - Some fish

| Carnivore | 3-4:1 | 12-30 hr | No |
| Omnivore | 6-9:1 | 24-72 hr | Limited |
