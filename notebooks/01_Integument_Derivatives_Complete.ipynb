## Part 1: Integument Structure

### What is Integument?

**Integument** = Outer covering of the body (skin + derivatives)
- Largest organ system
- First line of defense
- Multi-layered structure

### Two Primary Layers

#### **1. EPIDERMIS (Outer Layer)**
- **Origin:** Ectoderm
- **Type:** Stratified epithelium
- **Characteristics:** No blood vessels (avascular)
- **Cell types:** Keratinocytes, melanocytes, Langerhans cells, Merkel cells
- **Layers (thick skin, e.g., palm):**
  1. Stratum basale (basal layer) - mitotic, stem cells
  2. Stratum spinosum (spinous layer) - desmosomes
  3. Stratum granulosum (granular layer) - keratohyalin granules
  4. Stratum lucidum (clear layer) - only in thick skin
  5. Stratum corneum (horny layer) - dead, keratinized cells

#### **2. DERMIS (Inner Layer)**
- **Origin:** Mesoderm
- **Type:** Dense connective tissue
- **Characteristics:** Vascular, nerve-rich
- **Two zones:**
  1. Papillary layer (upper) - loose connective tissue, dermal papillae
  2. Reticular layer (lower) - dense irregular connective tissue, collagen/elastin

#### **3. HYPODERMIS (Subcutaneous Layer)**
- **Not technically part of skin**
- Loose connective tissue + adipose
- Anchors skin to underlying structures
- Insulation, energy storage

### Vertebrate Variations

**Fish:**
- Thin epidermis (few layers)
- Mucous glands abundant
- Dermis with scales

**Amphibians:**
- Thin, moist epidermis
- Many mucous + poison glands
- Highly vascularized (cutaneous respiration)

**Reptiles:**
- THICK, keratinized epidermis
- Few glands (mostly scent glands)
- Scales prevent water loss

**Birds:**
- Thin epidermis
- Few glands (uropygial/preen gland)
- Feathers for insulation

**Mammals:**
- Thick epidermis
- MANY glands (sweat, sebaceous, mammary)
- Hair for insulation

## Summary

### Integument Structure

**Two Primary Layers:**
1. **Epidermis** (Ectoderm): Stratified epithelium, avascular, keratinized
2. **Dermis** (Mesoderm): Connective tissue, vascular, nerve-rich

**Evolutionary Trends:**
- Keratinization: Minimal (fish/amphibian) → Extensive (reptile/mammal)
- Glands: Mucous (fish/amphibian) → Few (reptile) → Many (mammal)
- Water retention: Poor (amphibian) → Excellent (reptile)

### Functions (10+ roles)

1. **Protection** - Physical, chemical, biological
2. **Thermoregulation** - Sweat, hair, feathers, behavioral
3. **Sensory** - Mechanoreceptors, thermoreceptors, nociceptors
4. **Excretion** - Sweat (minor), salt glands
5. **Water Balance** - Retention vs absorption
6. **Vitamin D Synthesis** - UV-B → Vitamin D3
7. **Communication** - Color, scent, display
8. **Respiration** - Amphibians (30-100% cutaneous)
9. **Metabolic** - Fat storage, calcium reservoir
10. **Locomotion** - Friction, hydrodynamics

### Scales

**Fish (4 types, Dermal origin):**
- **Placoid** (sharks): Tooth-like denticles, homologous to teeth!
- **Ganoid** (gars): Diamond-shaped armor, ganoine layer
- **Cycloid** (salmon): Round, smooth, growth rings (age determination)
- **Ctenoid** (perch): Like cycloid + comb-like edge

**Reptiles (Epidermal, β-keratin):**
- Overlapping keratinized folds
- Periodic molting/shedding
- Some with osteoderms (bony plates underneath)
- Turtle scutes (epidermal over bony shell)

### Claws, Nails, Hooves

**Evolutionary Progression:**
Ancestral CLAW → Modified into:

**CLAW:**
- Curved, pointed, wraps around digit
- Reptiles, birds, most mammals
- Function: Climbing, digging, defense, prey capture
- Example: Cat (retractable), eagle talons

**NAIL:**
- Flat, broad, dorsal surface only
- Primates (humans, apes, monkeys)
- Derived from claws for precision grip
- Function: Fine manipulation, protection

**HOOF:**
- Thick capsule enclosing digit tip
- Ungulates (horses, cattle, deer)
- Function: Weight-bearing, running
- Types: Odd-toed (horse: 1) vs Even-toed (cow: 2)

### Hair (Mammals Only)

**Structure:**
- Keratinized filament from hair follicle
- Layers: Medulla, Cortex, Cuticle
- α-keratin (hard)
- Cyclic growth (anagen → catagen → telogen)

**Types:**
- Guard hair (long, protective)
- Underfur (short, insulating)
- Vibrissae (whiskers, sensory)

**Functions:**
- Insulation, camouflage, sensory, UV protection

**Modifications:**
- Quills (porcupine), Spines (hedgehog), Whiskers (cats)

### Feathers (Birds Only)

**Structure:**
- Complex branched from follicle
- Rachis → Barbs → Barbules → Hooklets
- β-keratin (lightweight)

**Five Types:**
1. **Contour** - Outer covering, flight (remiges, retrices)
2. **Down** - Fluffy insulation
3. **Semiplume** - Combination
4. **Filoplume** - Hair-like, sensory
5. **Bristle** - Stiff, sensory

**Functions:**
- Flight, insulation, waterproofing, display

**Maintenance:**
- Preening with uropygial gland oil

### Dentition

**Tooth Structure:**
- Enamel (hardest, ectoderm)
- Dentine (bulk, mesoderm)
- Pulp cavity (blood, nerves)
- Cementum (root covering)

**Mammalian Tooth Types (Heterodont):**
- **I** (Incisors): Cutting/gnawing
- **C** (Canines): Tearing/piercing
- **PM** (Premolars): Crushing
- **M** (Molars): Grinding

**Patterns:**
- **Homodont**: All similar (fish, reptiles)
- **Heterodont**: Differentiated (mammals)
- **Polyphyodont**: Continuous replacement (fish, reptiles)
- **Diphyodont**: Two sets (most mammals)

**Dental Formulas (examples):**
- Human: 2.1.2.3 / 2.1.2.3 = 32 (omnivore)
- Dog: 3.1.4.2 / 3.1.4.3 = 42 (carnivore)
- Cow: 0.0.3.3 / 4.0.3.3 = 32 (herbivore - no upper incisors!)

**Diet Adaptations:**
- Carnivore: Large canines, carnassials, reduced molars
- Herbivore: Reduced/no canines, large molars
- Omnivore: All tooth types present

### Key Comparisons

**Keratin Types:**
- α-keratin: Mammals (hair, nails, horns)
- β-keratin: Reptiles, Birds (scales, feathers, claws)

**Hair vs Feathers:**
- NOT homologous (convergent for insulation)
- Different keratin (α vs β)
- Different structure (simple vs complex branched)
- Different growth (continuous vs molting)

**Fish vs Reptile Scales:**
- Fish: Dermal (bony), embedded
- Reptile: Epidermal (keratin), surface

**Special Note:**
- Placoid scales (sharks) are HOMOLOGOUS to teeth!
- Same structure: enamel, dentine, pulp
