## Part 1: Classification of Receptors

### What are Receptors?

**Receptors** = Specialized cells or nerve endings that detect stimuli and convert them into electrical signals

### Three Classification Systems

#### **1. By Stimulus Type (Most Common)**

**PHOTORECEPTORS** - Light
- Rods (dim light, black & white)
- Cones (bright light, color)
- Location: Retina (eye)

**MECHANORECEPTORS** - Mechanical forces (pressure, vibration, stretch)
- Skin: Merkel, Meissner, Pacinian, Ruffini
- Ear: Hair cells (hearing, balance)
- Proprioceptors: Muscle spindles, Golgi tendon organs
- Baroreceptors: Blood pressure

**CHEMORECEPTORS** - Chemicals
- Taste buds (gustatory)
- Olfactory receptors (smell)
- Carotid/aortic bodies (O₂, CO₂, pH)

**THERMORECEPTORS** - Temperature
- Cold receptors (10-35°C)
- Warm receptors (30-45°C)

**NOCICEPTORS** - Pain (tissue damage)
- Mechanical, thermal, chemical pain

#### **2. By Location**

**EXTEROCEPTORS** - External environment
- Skin receptors (touch, pain, temperature)
- Special senses (vision, hearing, taste, smell)

**INTEROCEPTORS** - Internal environment
- Visceral receptors (stretch, pain in organs)
- Chemoreceptors (blood chemistry)
- Baroreceptors (blood pressure)

**PROPRIOCEPTORS** - Body position/movement
- Muscle spindles
- Golgi tendon organs
- Joint receptors

#### **3. By Structure**

**FREE NERVE ENDINGS** - Unencapsulated
- Pain, temperature, some touch
- Simple dendrites

**ENCAPSULATED** - Surrounded by connective tissue
- Pacinian corpuscles (deep pressure)
- Meissner's corpuscles (light touch)
- Ruffini endings (stretch)

**SPECIALIZED CELLS** - Non-neural origin
- Taste receptor cells
- Olfactory receptor cells
- Hair cells (ear)
- Photoreceptors (rods, cones)

## Summary

### Receptor Classification

**By Stimulus Type (5 categories):**
1. **Photoreceptors** - Light
2. **Mechanoreceptors** - Mechanical forces
3. **Chemoreceptors** - Chemicals
4. **Thermoreceptors** - Temperature
5. **Nociceptors** - Pain/tissue damage

### Visual Receptors (Eye)

**Photoreceptors in Retina:**

**RODS (120 million):**
- Peripheral retina (absent in fovea)
- Scotopic vision (dim light, night)
- Rhodopsin (peak 500 nm)
- Very high sensitivity, low acuity
- No color (monochromatic)

**CONES (6 million):**
- Concentrated in fovea
- Photopic vision (bright light, color)
- 3 types: S (blue), M (green), L (red)
- Low sensitivity, very high acuity
- Trichromatic color vision

**Key Features:**
- Fovea: Only cones, highest acuity
- Phototransduction: Light → HYPERPOLARIZATION (unusual!)
- Pathway: Retina → LGN → V1 (occipital cortex)

### Auditory Receptors (Ear)

**Hair Cells in Cochlea:**

**INNER HAIR CELLS (3,500):**
- Single row
- PRIMARY auditory receptors
- 95% of auditory nerve fibers
- Detect sound

**OUTER HAIR CELLS (12,000):**
- Three rows
- Cochlear AMPLIFIER (electromotility)
- Motor protein prestin
- Amplify 100-1000x

**Mechanoelectrical Transduction:**
- Sound → Basilar membrane vibrates
- Stereocilia bend (toward tallest = excitation)
- Tip links → K+ channels OPEN
- Depolarization → Glutamate release
- FASTEST transduction (microseconds!)
- Detect 0.3 nm movements!

**Frequency Coding:**
- Base: Narrow, stiff → HIGH freq (20 kHz)
- Apex: Wide, floppy → LOW freq (20 Hz)
- Tonotopic map maintained to cortex

### Chemoreceptors

**GUSTATORY (Taste):**
- 2,000-10,000 taste buds
- Five tastes: Sweet, Sour, Salty, Bitter, Umami
- Mechanisms: GPCRs (sweet, bitter, umami) or Ion channels (salty, sour)
- Myth busted: All tastes detected everywhere (no tongue map!)

**OLFACTORY (Smell):**
- 10-20 million olfactory neurons
- ~400 different GPCRs
- Sensitivity: Parts per trillion
- Neurons regenerate (30-60 days)
- UNIQUE: Bypasses thalamus (direct to cortex)

**INTERNAL (Respiratory Control):**

*Peripheral:*
- Carotid bodies: O₂, CO₂, pH (CN IX)
- Aortic bodies: O₂, CO₂, pH (CN X)

*Central (Most Important):*
- Medulla: CO₂ in CSF (NOT O₂!)
- 70-80% of respiratory drive

### Mechanoreceptors

**CUTANEOUS (Skin):**

*Rapidly Adapting (detect changes):*
- Meissner: Light touch, texture
- Pacinian: Deep pressure, vibration

*Slowly Adapting (sustained):*
- Merkel: Fine details, texture
- Ruffini: Skin stretch, pressure

**PROPRIOCEPTORS:**
- Muscle spindles: Muscle LENGTH (parallel)
- Golgi tendon organs: Muscle TENSION (series)
- Joint receptors: Joint angle

**VESTIBULAR (Balance):**
- Semicircular canals (3): Angular acceleration
- Utricle: Linear (horizontal)
- Saccule: Linear (vertical) + gravity
- Hair cells with otoliths (crystals)

**OTHER:**
- Baroreceptors: Blood pressure
- Visceral: Organ stretch

### Key Concepts

**Adequate Stimulus:**
- Each receptor most sensitive to ONE stimulus type

**Adaptation:**
- Rapidly adapting (phasic): Detect CHANGES
- Slowly adapting (tonic): Detect SUSTAINED

**Transduction:**
- Convert stimulus → electrical signal
- All use ion channels (directly or via cascade)

**Unusual Features:**
- Photoreceptors: Light HYPERPOLARIZES
- Olfaction: Bypasses thalamus
- Hair cells: Fastest transduction

You've mastered receptor classification, explored photoreceptors (rods vs cones, phototransduction), analyzed auditory hair cells (inner vs outer, mechanoelectrical transduction), studied chemoreceptors (taste, smell, internal), and examined mechanoreceptors (skin, proprioception, balance)!

**Key Takeaway:** Sensory receptors are specialized transducers converting specific stimuli into electrical signals.
